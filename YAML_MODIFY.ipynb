```python
#-------------------------------------------------------------------------------------------------------------
# The code to convert optuna from xlsx to config.yaml file (Only the parameters used in the project)
# It should be then pasted in config.yaml file to run the experiments
#-------------------------------------------------------------------------------------------------------------
import ast
import yaml
import pandas as pd

path_xlsx = "H:/My Drive/study/graduation_project/final/Code/project_files/TrafficManager/TrafficManager/SUMO-RL-TEST/SUMO-RL-PARAMS.xlsx"
path_yaml = "H:/My Drive/study/graduation_project/final/Code/project_files/TrafficManager/TrafficManager/SUMO-RL-TEST/config.yaml"
data = pd.read_excel(path_xlsx)
df = pd.DataFrame(data)
```

```python
# SUMO-RL Experiments
def create_yaml_sumo_rl(df):
    yaml_dict = {}

    for _, row in df.iterrows():
        algo = row["Algorithm"].lower()  # ppo or d3qn
        area = row["Area"].split()[0]  # Extract "Area2" or "Area3"
        area = "Mosheer" if "Mosheer" in row["Area"] else "Stefano"
        reward_type = row["Reward"].lower()  # literature or proposed
        key_name = f"Sumorl_{area}_{reward_type}_reward_{algo}"

        entry = ast.literal_eval(row["Best Parameters"])
        entry["best_value"] = row["Best Value"]

        yaml_dict.setdefault(key_name, entry)

    return yaml_dict

# Create the structured dictionary
yaml_data = create_yaml_sumo_rl(df)

# Dump to YAML
yaml_output = yaml.dump(yaml_data, sort_keys=False)

# Save to file
with open(path_yaml,'w') as f:
    f.write(yaml_output)

print(yaml_output)
```

```python
# Enviromental Experiments
def create_yaml_structure_env(df):
    yaml_dict = {"PPO": {}, "D3QN": {}}

    for _, row in df.iterrows():
        algo = row["Algorithm"]
        area = row["Area"].split()[0]  # Extract "Area2" or "Area3"
        area = "Mosheer" if "Mosheer" in row["Area"] else "Stefano"
        env = row["Environment Type"]
        reward_type = row["Reward"].lower()  # literature or proposed
        key_name = f"{area}_{env}_{reward_type}_reward"

        entry = ast.literal_eval(row["Best Parameters"])
        entry["best_value"] = row["Best Value"]

        yaml_dict[algo].setdefault(key_name, entry)

    return yaml_dict

# Create the structured dictionary
yaml_data = create_yaml_structure_env(df)

# Dump to YAML
yaml_output = yaml.dump(yaml_data, sort_keys=False)

# Save to file
with open(path_yaml) as f:
    f.write(yaml_output)

print(yaml_output)
```
